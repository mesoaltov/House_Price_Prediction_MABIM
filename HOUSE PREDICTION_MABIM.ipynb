```python
import pandas as pd
```

```python
df_rumah = pd.read_csv('/content/data.csv')

df_rumah
```

```python
df_rumah_bedroom=df_rumah[['bedrooms']]

df_rumah_bedroom
```

```python
# jumlah_kamar.columns = ['Jumlah Kamar', 'Jumlah Rumah']

# jumlah_kamar
```

```python
df_rumah_bedroom ['Total Rumah'] = df_rumah_bedroom.groupby('bedrooms')['bedrooms'].transform('count')

df_rumah_bedroom
```

```python
df_rumah_bedroom = df_rumah_bedroom.drop_duplicates().sort_values(by='bedrooms')

df_rumah_bedroom
```

```python
import matplotlib.pyplot as plt

df_rumah_bedroom.plot(
    x='bedrooms',
    y='Total Rumah',
    kind='bar',
    title='Jumlah Rumah Berdasarkan Jumlah Kamar Tidur',
    xlabel='Jumlah Kamar Tidur',
    ylabel='Jumlah Rumah'
)

plt.show()
```
